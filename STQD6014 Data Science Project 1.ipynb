PART 2 LISTS, DICTIONARIES AND LOOPING

```python
# Dictionary storing info about 5 cities
cities = {
    "hanoi": {
        "country": "vietnam",
        "population_in_millions": 5.6,
        "landmarks": ["train street", "hoa lo prison", "mega grand world hanoi"],
        "dishes": ["banh mi", "pho", "vietnamese spring rolls"]
    },
    "phuket": {
        "country": "thailand",
        "population_in_millions": 0.5,
        "landmarks": ["kata beach", "old phuket town", "big buddha"],
        "dishes": ["pad thai", "tom yum goong", "pad krapaw"]
    },
    "vientiane": {
        "country": "laos",
        "population_in_millions": 1.0,
        "landmarks": ["patuxay monument", "pha that luang", "wat sisaket"],
        "dishes": ["sai oua", "nam khao", "khao poon nam paa"]
    },
    "phnom penh": {
        "country": "cambodia",
        "population_in_millions": 3.5,
        "landmarks": ["royal palace phnom penh", "silver pagoda", "independence monument"],
        "dishes": ["amok", "kuy teav", "lap khmer"]
    },
    "naypyidaw": {
        "country": "myanmar",
        "population_in_millions": 0.9,
        "landmarks": ["uppatasanti pagoda", "myoma market", "water fountain garden"],
        "dishes": ["nan gyi thoke", "burmese tofu", "mohinga"]
    }
}

# Printing information about the cities using loops
for city, info in cities.items():
    print("\n" + city.upper())
    print("Country: " + info['country'].title())
    print("Population: " + str(info['population_in_millions']) + " million")

    print("  Famous Landmarks:")
    for landmark in info["landmarks"]:
        print("\t- " + landmark.title())

    print("  Local Dishes:")
    for dish in info["dishes"]:
        print("\t- " + dish.title())
```

PART 5 NUMPY

```python
import numpy as np
```

```python
# Two arrays each of sizes: single 1d array, 3x3, 2x4, 4x2 and 5x5
A = np.random.randn(5)
B = np.random.randn(5)
C = np.random.randn(3, 3)
D = np.random.randn(3, 3)
E = np.random.randn(2, 4)
F = np.random.randn(2, 4)
G = np.random.randn(4, 2)
H = np.random.randn(4, 2)
I = np.random.randn(5, 5)
J = np.random.randn(5, 5)
```

```python
# Operation on single array
# Absolute value of A
np.abs(A)
```

```python
# B^2
np.square(B)
```

```python
# Sign of each element in C
np.sign(C)
```

```python
# exp^x where x is each element in D
np.exp(D)
```

```python
# Compute the floor of each element (the largest integer less than or equal to each element)
np.floor(E)
```

```python
# Return fractional and integral parts of array as a separate array
np.modf(F)
```

```python
# A multiply by 3
A*3
```

```python
# B divide by 2
B/2
```

```python
# Maximum between E and F
np.maximum(E,F)
```

```python
# C multiply by D
np.multiply(C,D)
```

```python
# A + H
np.add(G,H)
```

```python
# I divide J
np.divide(I,J)
```

```python
# Element-wise modulus (remainder of division)
np.mod(I,J)
```

```python
# Sum
np.sum(I)
```

```python
# Cumulative sum of elements starting from 0
np.cumsum(J)
```

```python
# Cumulative product of elements starting from 1
np.cumprod(F)
```

```python
# Variance
np.var(G)
```

```python
# Standard Deviation
np.std(E)
```

```python
# Mean
np.mean(H)
```

```python
from numpy.linalg import *
```

```python
# Diagonal element
np.diag(C)
```

```python
# Compute the sum of the diagonal elements
np.trace(D)
```

```python
# Compute the matrix determinant
np.linalg.det(J)
```

```python
# Compute eigen valus and eigen vectors
values, vectors = np.linalg.eig(I)

print("Eigenvalues:")
print(values)

print("Eigenvectors:")
print(vectors)
```

```python
# Inverse matrix
np.linalg.inv(D)
```

```python
# Multiplication matrix (2 x 4 and 4 x 2)
np.dot(E, H)
```

```python
# Multiplication matrix (4 x 2 and 2 x 4)
np.dot(G, F)
```

```python
# Multiplication matrix (5 x 5 and 5 x 5)
np.dot(I, J)
```

```python
# Multiplication matrix (3 x 3 and 3 x 3)
np.dot(C, D)
```

```python
# Solve the linear system Ix = B for x
np.linalg.solve(I, B)
```
